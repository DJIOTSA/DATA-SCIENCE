# Introduction to data visualization

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
seattle_weather = pd.read_csv("seattle_weather.csv")
seattle_weather
```

```python
austin_weather = pd.read_csv("austin_weather.csv")
austin_weather
```

## multiple gragh

```python
fig, ax = plt.subplots()
ax.plot(austin_weather['DATE'], austin_weather['MLY-TAVG-NORMAL'], label='Average Temperature', marker='o', color='blue')
ax.plot(austin_weather['DATE'], austin_weather['MLY-TMIN-PRBOCC-LSTH024'], label='Average Minimum Temperature', marker='v', color='r', linestyle='--')
ax.plot(austin_weather['DATE'], austin_weather['MLY-TMIN-PRBOCC-LSTH036'], label='Average Minimum Temperature 36', marker='v', color='yellow', linestyle=None
        )

ax.set_xlabel('Date (months)')
ax.set_ylabel('Average Temperature')
ax.set_title('Austin Weather Data')
fig.set_size_inches(10, 6)

ax.legend()
plt.show()
```

## Time series

```python
climate_change = pd.read_csv("climate_change.csv", parse_dates=['date'], index_col='date')
climate_change
```

```python
fig, ax = plt.subplots()
ax.plot(climate_change.index, climate_change['relative_temp'])
ax.set_xlabel('Time')
ax.set_ylabel('Relative temperature (Celsius)')
plt.show()
```

```python
fig, ax = plt.subplots()
ax.plot(climate_change.index, climate_change['co2'], color='blue',  label='CO2 Levels')
ax.set_xlabel('Time')
ax.set_ylabel('CO2 levels (ppm)', color='blue')
ax2 = ax.twinx()
ax2.plot(climate_change.index, climate_change['relative_temp'], color='red', label='Relative Temperature')
ax2.set_ylabel('Relative temperature (Celsius)', color='red')
plt.title('CO2 Levels and Relative Temperature Over Time')
fig.set_size_inches(10, 6)
ax.legend(loc='upper left')
ax2.legend(loc='lower right')
ax.tick_params(axis='y', colors='blue')
ax2.tick_params(axis='y', colors='red')
plt.show()
```

```python
def plot_timeseries(axes, x, y, xlabel, ylabel, color='blue'):
    axes.plot(x, y, color=color)
    axes.set_xlabel(xlabel)
    axes.set_ylabel(ylabel, color=color)
    axes.tick_params(axis='y', colors=color)

fig, ax = plt.subplots()
fig.set_size_inches(10, 6)
plot_timeseries(ax, climate_change.index, climate_change['co2'],'Time', 'CO2 levels (ppm)', color='blue')
ax2 = ax.twinx()
plot_timeseries(ax2, climate_change.index, climate_change['relative_temp'], 'Time', 'Relative temperature (Celsius)', color='red')
ax.legend(['CO2 Levels'], loc='upper left')
ax2.legend(['Relative Temperature'], loc='lower right')
plt.title('CO2 Levels and Relative Temperature Over Time')

ax2.annotate('>1 degree C', xy=(pd.Timestamp('2015-01-01'), 1), xytext=(pd.Timestamp('2005-01-01'), -0.25), arrowprops={"arrowstyle": "->", 'color':'green'})
plt.grid()
plt.show()
```

# Quantitative bar chart

```python
medals_by_country = pd.read_csv("medals_by_country_2016.csv", index_col=0)
medals_by_country
```

```python
fig, ax = plt.subplots()
ax.bar(medals_by_country.index, medals_by_country['Gold'], label='Gold', color='gold')
ax.bar(medals_by_country.index, medals_by_country['Silver'], bottom=medals_by_country['Gold'], label='Silver', color='silver')
ax.bar(medals_by_country.index, medals_by_country['Bronze'], bottom=medals_by_country['Gold'] + medals_by_country['Silver'], label='Bronze', color='brown')
ax.set_xticklabels(medals_by_country.index, fontsize=14, rotation=90)
ax.set_ylabel('Total Medals')
ax.set_title('Total Medals by Country in 2016 Olympics')
ax.legend()
fig.set_size_inches(12, 6)
plt.show()
```

```python
# Histogram
```

```python
summer = pd.read_csv("summer2016.csv")
summer
```

```python
fig , ax = plt.subplots()
fig.set_size_inches(16, 9)

ax.hist(summer[(summer['Sport'] == 'Rowing') & (summer['Sex'] == 'M')]['Height'], color='skyblue', edgecolor='black', histtype='step', bins=5)
ax.hist(summer[(summer['Sport'] == 'Gymnastics') & (summer['Sex'] == 'M')]['Height'], color='lightgreen', edgecolor='black', alpha=0.7, histtype='stepfilled', bins=5)
ax.set_xlabel('Height (cm)')
ax.set_ylabel('Number of Athletes')
ax.set_title('Height Distribution of Male Athletes')
plt.show()
```

```python
fig , ax = plt.subplots()
fig.set_size_inches(16, 9)
mens_rowing = summer[(summer['Sport'] == 'Rowing') & (summer['Sex'] == 'M')]
mens_gymnastics = summer[(summer['Sport'] == 'Gymnastics') & (summer['Sex'] == 'M')]
womens_rowing = summer[(summer['Sport'] == 'Rowing') & (summer['Sex'] == 'F')]
womens_gymnastics = summer[(summer['Sport'] == 'Gymnastics') & (summer['Sex'] == 'F')]

ax.hist(mens_rowing['Height'], color='red', edgecolor='blue', histtype='step', bins=[150,160, 170, 180,190,200, 210], label='Rowing')
ax.hist(mens_gymnastics['Height'], color='lightgreen', edgecolor='black', alpha=0.7, histtype='stepfilled', bins=[150,160, 170, 180,190,200, 210], label='Gymnastics')
ax.set_xlabel('Height (cm)')
ax.set_ylabel('Number of Athletes')
ax.set_title('Height Distribution of Male Athletes')
ax.legend()
plt.show()
```

# Adding error bar in bad plot

```python
fig, ax = plt.subplots()

ax.bar('Rowing', mens_rowing['Height'].mean(), color='red', label='Rowing', yerr=mens_rowing['Height'].std())
ax.bar('Gymnastics', mens_gymnastics['Height'].mean(), color='lightgreen', label='Gymnastics', alpha=0.7, yerr=mens_gymnastics['Height'].std())

ax.set_xlabel('Height (cm)')
ax.set_ylabel('Number of Athletes')
ax.set_title('Height Distribution of Male Athletes')
ax.legend()
plt.show()
```
