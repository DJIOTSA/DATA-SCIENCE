## Statistic: is the process or the study of collecting data, analyzing data and drawing results.

## Summary statistic is a fact or a summary about a data.

## Why statistic is the foundation of data science: Because without it data science will just be coding and number, not real insight. Data science gives us tools to handle and automate large, messy data while Statistics gives us the “thinking rules” to avoid fooling ourselves and to make sure what we see is real.

### 1. Understanding data properly

Stats tells us how to summarize data (mean, median, variance) and see patterns.

Without this, machine learning would just be a "black box."

### 2. Avoiding wrong conclusions

Example: Correlation ≠ causation. Statistics helps us test whether results are significant or just random noise.

### 3. Designing experiments

A/B testing (like testing two website designs) comes directly from statistical hypothesis testing.

### 4. Making predictions reliable

Stats gives us confidence intervals, p-values, distributions, etc., so we know how much to trust a prediction.

### 5. Interpreting machine learning models

Even ML uses concepts like probability distributions, loss functions, regression, Bayes’ theorem—all from statistics.

# Mesure of the center of the data: mean, median, mode

```python
import pandas as pd
import numpy as np
import statistics

amir_deals = pd.read_csv('amir_deals.csv')
food_consumption = pd.read_csv('food_consumption.csv')
world_happiness = pd.read_csv('world_happiness.csv')
```

```python
# Mean: average value
print(amir_deals.head())

print("amir_deals mean price:",np.mean(amir_deals['amount']))
```

```python
# Median: the middle value
print('amir_deals median:', np.median(amir_deals['amount']))
```

```python
# Mode: the most frequent value
print('amir_deals mode:', statistics.mode(amir_deals['amount']))
```

```python
help(statistics)
```

## Mesure of spread: variance, std

```python
help(np.quantile)
```

```python
# Variance: average distance of each data points from the data's mean

print('statistics Variance:', statistics.variance(amir_deals['amount']))
print('np Variance:', np.var(amir_deals['amount']))
print('statistic Std:', statistics.stdev(amir_deals['amount']))
print('np Std:', np.std(amir_deals['amount']))

# quantile of 0.5 = median
print('statistics.median:', statistics.median(amir_deals['amount']))
print('np.median:', np.median(amir_deals['amount']))
print('statistics.quantiles:', statistics.quantiles(amir_deals['amount']))
print('np qantile 0,5:', np.quantile(amir_deals['amount'],  0.5))

# quatiles
print('np quartile:', np.quantile(amir_deals['amount'], [0, 0.25, 0.5, 0.75, 1]))




```

```python
# boxplot use quartiles
import matplotlib.pyplot as plt

plt.boxplot(amir_deals['amount'])
plt.grid(visible=True, alpha=0.3, fillstyle='full' )
plt.show()
```

```python
help(plt.grid)
```

```python
# quantiles linspace
print('test:', np.quantile(amir_deals['amount'], [0, 0.2, 0.4, 0.6, 0.8, 1]))
print('test:', np.quantile(amir_deals['amount'], np.linspace(0,1, 5)))
```

```python
# interquatile range
plt.boxplot(amir_deals['amount'])
plt.show()

iqr = np.quantile(amir_deals['amount'], 0.75) - np.quantile(amir_deals['amount'], 0.25)
print({"iqr": iqr})
```

```python
# interquartile range 
from scipy.stats import iqr
iqr2 =  iqr(amir_deals['amount'])
print({iqr2})
```

## Outliers: data that are possibly different from others:

***Q1: quantile 0.25***

***Q3: quantile 0.75***

***IQR: interquartile range***

## Outlier data are data which are: data <  Q1 -1.5 * IQR or  data > Q3+1.5 * IQR


```python
# Lets find outlier of amir_deal num_users

# interquartile
iqr3 = iqr(world_happiness['life_exp'])
lower_threhold = np.quantile(world_happiness['life_exp'], 0.25) -1.5*iqr3;
upper_threhold = np.quantile(world_happiness['life_exp'], 0.75) + 1.5*iqr3;
outliers_data = world_happiness[(world_happiness['life_exp']< lower_threhold) | (world_happiness['life_exp'] > upper_threhold) ]
print({lower_threhold, upper_threhold, iqr3})
outliers_data
```
