# Uniform distribution in python

```python
from scipy.stats import uniform

# P(waiting time >= 7) = uniform.cdf(7,0,12)
# Where 12 represent the uniform time
uniform.cdf(7,0,12)
```

```python
help(uniform.cdf)
```

```python
# P(4 <= waiting time <= 7)= P(waiting time >= 7) - P(waiting time >=4)
uniform.cdf(7,0,12) - uniform.cdf(4,0,12)
```
