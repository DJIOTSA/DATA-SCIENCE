# Binomial distribution distribution

## 1. Fixed number of trials
## 2. Two possible outcome par trial
## 3. Each trial is indepent
## If n is the number of trials (integer) and p the probability of success then:
***mean = n * p***

***variance = n * p * (1 - p)*** 

## randomly flipping one or multiples coins 1 or many time

```python
# filling a coin with 50% of change ot succeed 1 times
from scipy.stats import binom

number_of_coins = 1
probability_of_success = 0.5
number_of_time= 1

binom.rvs(number_of_coins, probability_of_success, size=number_of_time)
```

```python
# filling a coin with 50% of change ot succeed 10 times
from scipy.stats import binom

number_of_coins = 1
probability_of_success = 0.5
number_of_time= 10

binom.rvs(number_of_coins, probability_of_success, size=number_of_time)
```

```python
# filling 10 coins with 2% of change ot succeed 1 times
from scipy.stats import binom

number_of_coins = 10
probability_of_success = 0.02
number_of_time= 1

binom.rvs(number_of_coins, probability_of_success, size=number_of_time)
```

```python
# filling 7 coins with 35% of change of succeed 4 times
from scipy.stats import binom

number_of_coins = 7
probability_of_success = 0.35
number_of_time= 3

binom.rvs(number_of_coins, probability_of_success, size=number_of_time)
```

# getting exactly 4 success out of 7 trials (pmf: probability mass function)

```python
probability_of_success = 0.5
number_of_success = 4,
number_of_trials = 70

binom.pmf(number_of_success, number_of_trials, probability_of_success)
```
