```python
import pandas as pd 
import numpy as np 
import seaborn as sns
import matplotlib.pyplot as plt 
```

```python
# dataset
dataset = pd.read_csv('food_consumption.csv')
print(dataset.head())

# get consumption per country
consumption = dataset.groupby(by='country', as_index=False)['consumption'].mean()
consumption

data = {
    'Argentina': np.random.randint(size=50, low=50, high=2000, ),
    'Australia': np.random.randint(size=50, low=100, high=2500, ),
}

sns.scatterplot(x='Argentina', y='Australia', data=data, )
plt.show()
```
